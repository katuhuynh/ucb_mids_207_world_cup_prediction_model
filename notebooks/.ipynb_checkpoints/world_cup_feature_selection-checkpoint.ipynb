# FIFA Men's World Cup Feature Selection

This notebook builds a feature table for predicting FIFA Men's World Cup match outcomes.

## Selected sources

- `matches.csv`: primary match table and target
- `teams.csv`: team identity and confederation
- `eloratings.csv`: latest rating available on or before kickoff
- `tournaments.csv`: year, host country, and tournament size

## Excluded from the first model

- `results.csv`: mostly non-World-Cup matches; Elo already summarizes broader international strength
- `goalscorers.csv`: scorer identity is not necessary for the initial outcome model
- `shootouts.csv`: contains post-match information and would leak the result
- `tournament_standings.csv`: final position is post-tournament information unless shifted to prior tournaments

Team names are retained in the clean table rather than being dropped before preprocessing.

```python
import pandas as pd

features = pd.read_csv(
    "world_cup_selected_features.csv",
    parse_dates=["match_date"]
)

print(features.shape)
print(features["year"].min(), features["year"].max())
display(features.head())
```

## Feature groups

**Team identity**
- Home and away team names, codes, regions, and confederations

**Pregame Elo**
- Home Elo, away Elo, difference, absolute difference, change, rating age, and missingness

**Prior World Cup history**
- Prior matches and tournament appearances
- Wins, draws, losses, points, and goals
- Win rate, draw rate, points per match, and goal rates
- Home-minus-away matchup differences

**Match context**
- Group stage or knockout stage
- Stage name and group
- Host-team indicators
- Same-confederation matchup
- Number of participating teams

All rolling historical features exclude the current match.

```python
identity_features = [
    "home_team_name", "away_team_name",
    "home_team_code", "away_team_code",
    "home_confederation_code", "away_confederation_code",
]

elo_features = [
    "home_elo", "away_elo", "elo_diff", "elo_abs_diff",
    "home_elo_change", "away_elo_change",
    "home_elo_age_days", "away_elo_age_days",
    "home_elo_missing", "away_elo_missing",
]

context_features = [
    "stage_name", "group_name", "group_stage", "knockout_stage",
    "home_is_host", "away_is_host", "same_confederation", "count_teams",
]

history_features = [c for c in features.columns if "prior_wc_" in c]

model_features = identity_features + elo_features + context_features + history_features
X = features[model_features]
y = features["target"].astype(int)

print("Selected features:", len(model_features))
print("Rows:", len(X))
```

```python
# Tournament-based split for honest evaluation.
# Test data will be from Worldcup 2026
train = features[features["year"] <= 2018]
validation = features[features["year"] == 2022]

print("Train:", train.shape)
print("Validation:", validation.shape)
```

## Modeling recommendation

For XGBoost, one-hot encode the categorical team and context columns, median-impute numeric missing values, and preserve the Elo missingness indicators.

A tournament-based split is preferred over a random split. Use 2018 for validation and 2022 as the final test set.

The current target treats a penalty-shootout winner as the match winner because `matches.csv` records the advancing team as the result. For a regulation-time 1X2 target, shootout matches should instead be labeled as draws.
